```python
# Retail Sales Data Engineering Pipeline
# Bronze Layer - Raw Data Ingestion
# Customer data
customer_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/retail_practice/default/retail_sales_data/customers.csv")

# Product data
product_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/retail_practice/default/retail_sales_data/products.csv")

# Order data
order_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/retail_practice/default/retail_sales_data/orders.csv")

customer_df.write.format("delta").mode("overwrite").saveAsTable(
    "retail_practice.default.bronze_customers"
)

product_df.write.format("delta").mode("overwrite").saveAsTable(
    "retail_practice.default.bronze_products"
)

order_df.write.format("delta").mode("overwrite").saveAsTable(
    "retail_practice.default.bronze_orders"
)

print("Customers:", customer_df.count())
print("Products:", product_df.count())
print("Orders:", order_df.count())

print("Bronze ingestion completed successfully!")
```
